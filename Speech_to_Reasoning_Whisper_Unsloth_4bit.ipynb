## 1. Speech Recognition Using Whisper

Whisper converts audio input into text. The generated transcription is then passed to the reasoning model for generating intelligent responses.

````python
whisper_model = whisper.load_model("base")

print("Whisper model loaded successfully")
````

## 2. Loading Quantized LLM for Reasoning

This step loads a memory-efficient 4-bit quantized Large Language Model using Hugging Face Transformers. The quantized model reduces GPU memory usage while maintaining effective reasoning ability.

````python
model_name = "unsloth/Llama-3.2-3B-Instruct-unsloth-bnb-4bit"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto"
)

if tokenizer.pad_token_id is None:
    tokenizer.pad_token = tokenizer.eos_token

model.eval()

# Generation length is controlled by max_new_tokens
model.generation_config.max_length = None

print("Model:", model_name)
print("Quantized LLM loaded successfully")
````

## 4. Core Pipeline Functions

Whisper converts audio input into text. The transcribed text is then passed to the quantized LLM for reasoning and response generation.

````python
def generate_response(user_text, max_new_tokens=180):

    messages = [
        {
            "role": "system",
            "content": (
                "You are a helpful AI assistant. "
                "Answer clearly and concisely. "
                "If you are uncertain, say that you are uncertain."
            ),
        },
        {
            "role": "user",
            "content": user_text.strip(),
        },
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    ).to(model.device)

    prompt_length = inputs["input_ids"].shape[-1]

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.pad_token_id
            if tokenizer.pad_token_id is not None
            else tokenizer.eos_token_id,
        )

    new_tokens = outputs[0][prompt_length:]

    response = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    return response
````

````python
def transcribe_audio(audio_path):

    result = whisper_model.transcribe(
        audio_path,
        language="en",
        fp16=torch.cuda.is_available()
    )

    text = result["text"].strip()

    if not text:
        raise ValueError("No speech could be transcribed from the audio.")

    return text
````

## 5. End-to-End Speech-to-Reasoning Pipeline

The following function integrates the complete workflow into a single callable pipeline.

It performs:

**Audio Input → Whisper ASR → Text Transcription → Quantized LLM → Generated Response**

In addition to the generated output, the pipeline records transcription latency, LLM generation latency, total response time, and peak GPU memory usage. These measurements help evaluate the practical efficiency of the system.

````python
import time


def speech_to_reasoning(audio_path):

    # Reset GPU peak-memory statistics for this run
    if torch.cuda.is_available():
        torch.cuda.reset_peak_memory_stats()
        torch.cuda.synchronize()

    total_start = time.perf_counter()

    # -----------------------------
    # Step 1: Speech Recognition
    # -----------------------------
    asr_start = time.perf_counter()

    transcription = transcribe_audio(audio_path)

    if torch.cuda.is_available():
        torch.cuda.synchronize()

    asr_time = time.perf_counter() - asr_start

    # -----------------------------
    # Step 2: LLM Response Generation
    # -----------------------------
    llm_start = time.perf_counter()

    response = generate_response(transcription)

    if torch.cuda.is_available():
        torch.cuda.synchronize()

    llm_time = time.perf_counter() - llm_start

    total_time = time.perf_counter() - total_start

    # -----------------------------
    # GPU Memory Measurement
    # -----------------------------
    if torch.cuda.is_available():
        peak_memory_gb = (
            torch.cuda.max_memory_allocated() / (1024 ** 3)
        )
    else:
        peak_memory_gb = None

    return {
        "transcription": transcription,
        "response": response,
        "asr_time_seconds": round(asr_time, 2),
        "llm_time_seconds": round(llm_time, 2),
        "total_time_seconds": round(total_time, 2),
        "peak_gpu_memory_gb": (
            round(peak_memory_gb, 2)
            if peak_memory_gb is not None
            else "CPU execution"
        )
    }


print("End-to-end Speech-to-Reasoning pipeline ready")
````

### Automated Evaluation

The following evaluation runs the complete Speech-to-Reasoning pipeline on each test audio sample.

For every sample, the system records:

- Whisper transcription
- Word Error Rate (WER)
- ASR processing time
- LLM generation time
- Total end-to-end latency
- Peak allocated GPU memory
- Generated AI response

Lower WER values indicate better speech-recognition accuracy. A WER of `0.00` means that the Whisper transcription exactly matches the reference sentence after text normalization.

````python
import pandas as pd
import re
from jiwer import wer


def normalize_text(text):
    """
    Normalize text before calculating Word Error Rate.
    """
    text = text.lower().strip()
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text)
    return text


evaluation_results = []

for filename, reference in reference_transcripts.items():

    print(f"\nEvaluating: {filename}")

    result = speech_to_reasoning(filename)

    reference_clean = normalize_text(reference)
    transcription_clean = normalize_text(result["transcription"])

    word_error_rate = wer(
        reference_clean,
        transcription_clean
    )

    evaluation_results.append({
        "Audio File": filename,
        "Reference Transcript": reference,
        "Whisper Transcript": result["transcription"],
        "WER": round(word_error_rate, 3),
        "ASR Time (s)": result["asr_time_seconds"],
        "LLM Time (s)": result["llm_time_seconds"],
        "Total Time (s)": result["total_time_seconds"],
        "Peak GPU Memory (GB)": result["peak_gpu_memory_gb"],
        "AI Response": result["response"]
    })


evaluation_df = pd.DataFrame(evaluation_results)

print("\nEvaluation completed successfully.")
evaluation_df
````

````python
print("=== SYSTEM EVALUATION SUMMARY ===")

print(f"Number of test samples: {len(evaluation_df)}")
print(f"Average WER: {evaluation_df['WER'].mean():.3f}")
print(f"Average ASR time: {evaluation_df['ASR Time (s)'].mean():.2f} seconds")
print(f"Average LLM generation time: {evaluation_df['LLM Time (s)'].mean():.2f} seconds")
print(f"Average total response time: {evaluation_df['Total Time (s)'].mean():.2f} seconds")
print(
    f"Maximum peak GPU memory: "
    f"{evaluation_df['Peak GPU Memory (GB)'].max():.2f} GB"
)
````

### Evaluation Results and Discussion

The end-to-end Speech-to-Reasoning system was evaluated using three independently recorded voice queries covering general artificial intelligence, machine learning, and renewable energy topics.

The Whisper ASR component achieved very low transcription error across the test set. Two of the three audio samples produced exact normalized transcriptions, while one sample contained a minor word-level recognition error.

Speech recognition was considerably faster than language-model generation. ASR processing required less than one second per sample, whereas the quantized LLM required substantially more time to generate complete responses. Therefore, LLM inference represented the main source of end-to-end latency in the current implementation.

Peak allocated GPU memory remained stable across the evaluated samples, demonstrating the practical memory efficiency of the 4-bit quantized Llama 3.2 model on the Google Colab Tesla T4 environment.

The generated responses were relevant to the corresponding transcribed questions in the tested examples. However, this small evaluation set is intended to demonstrate pipeline functionality rather than establish general model accuracy. A larger and more diverse speech dataset would be required for a comprehensive assessment of ASR robustness and response quality.

## 7. Interactive Gradio Application

To make the Speech-to-Reasoning system accessible to users without requiring direct interaction with notebook code, an interactive Gradio interface is added.

The application allows a user to:

- Record speech directly from a microphone or upload an audio file.
- Convert the spoken query into text using OpenAI Whisper.
- Pass the transcription to the 4-bit quantized Llama 3.2 model.
- Display the generated AI response.
- Report ASR latency, LLM generation latency, total response time, and peak allocated GPU memory.

> **Note:** Generated responses may contain errors and should not be relied upon for professional, medical, legal, or other high-stakes advice.

````python
!pip install -q gradio
````

````python
def gradio_speech_to_reasoning(audio_path):

    if audio_path is None:
        return (
            "No audio provided.",
            "Please record or upload an audio file first.",
            "No performance data available."
        )

    try:
        result = speech_to_reasoning(audio_path)

        metrics = (
            f"**ASR Time:** {result['asr_time_seconds']} seconds  \n"
            f"**LLM Generation Time:** {result['llm_time_seconds']} seconds  \n"
            f"**Total Response Time:** {result['total_time_seconds']} seconds  \n"
            f"**Peak Allocated GPU Memory:** {result['peak_gpu_memory_gb']} GB"
        )

        return (
            result["transcription"],
            result["response"],
            metrics
        )

    except Exception as error:
        return (
            "Processing failed.",
            f"Error: {str(error)}",
            "No performance data available."
        )


print("Gradio application function ready")
````

````python
import gradio as gr

with gr.Blocks(title="Speech-to-Reasoning AI Assistant") as demo:

    gr.Markdown(
        """
        # 🎙️ Speech-to-Reasoning AI Assistant

        Speak or upload an audio question.

        **Pipeline:**
        Whisper ASR → Text Transcription → Unsloth Llama 3.2 3B Dynamic 4-bit → AI Response

        > Educational AI demonstration. Generated responses may contain errors.
        """
    )

    audio_input = gr.Audio(
        sources=["microphone", "upload"],
        type="filepath",
        label="Record or Upload Audio"
    )

    submit_button = gr.Button(
        "Generate Response",
        variant="primary"
    )

    transcription_output = gr.Textbox(
        label="Whisper Transcription",
        lines=3
    )

    gr.Markdown("### AI Response")
    response_output = gr.Markdown()

    gr.Markdown("### System Performance")
    performance_output = gr.Markdown()

    submit_button.click(
        fn=gradio_speech_to_reasoning,
        inputs=audio_input,
        outputs=[
            transcription_output,
            response_output,
            performance_output
        ]
    )

demo.launch(share=True)
````

## 8. Limitations

Although the system successfully demonstrates an end-to-end Speech-to-Reasoning workflow, several limitations should be considered.

- Speech-recognition quality may decrease with background noise, strong accents, poor microphones, or longer audio inputs.
- The evaluation used a small set of manually recorded queries and therefore does not represent a comprehensive benchmark.
- Large Language Models can generate incomplete or factually incorrect responses even when the speech transcription is accurate.
- LLM inference contributes most of the end-to-end response latency in the current implementation.
- The temporary Gradio sharing URL depends on the active Google Colab runtime and is not intended for permanent deployment.
- The application is designed as an educational AI prototype and should not be used as a substitute for professional advice in high-stakes domains.
